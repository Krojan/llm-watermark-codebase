```python
import json

def read_json(filepath):
    with open(filepath, "r", encoding="utf-8") as f:
        dataset = json.load(f)
    return dataset

def write_json(filename, json_data):
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(json_data, f, indent=2, ensure_ascii=False)
```

```python

import json
import sys
from pathlib import Path

input_folder = Path('/Users/rojankarki/Projects/watermark-llm-code-quality/watermark-llm-test-dataset/dataset/promptDescription-1XX')
output_file = 'promptDescription-0XX.json'


json_files = sorted(input_folder.glob("*.json"))
print(len(json_files))

if not json_files:
    print(f"No .json files found in '{input_folder}'.")

merged = []
skipped = []

for jf in json_files:
    try:
        with open(jf, "r", encoding="utf-8") as f:
            data = json.load(f)
    except json.JSONDecodeError as e:
        skipped.append((jf.name, f"invalid JSON: {e}"))
        continue
    except Exception as e:
        skipped.append((jf.name, str(e)))
        continue

    tasks = data.get("tasks")
    if not tasks or not isinstance(tasks, list) or len(tasks) == 0:
        skipped.append((jf.name, "no 'tasks' list or it's empty"))
        continue

    merged.append(tasks[0])

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(merged, f, indent=2, ensure_ascii=False)

print(f"Processed {len(json_files)} file(s).")
print(f"Merged {len(merged)} task(s) into '{output_file}'.")

if skipped:
    print(f"\nSkipped {len(skipped)} file(s):")
    for name, reason in skipped:
        print(f"  - {name}: {reason}")
```
